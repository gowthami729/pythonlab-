#Select Two Columns and Plot Scatter Graph

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Age": [18, 19, 20, 18, 21, 19, 20, 21, 18, 22],
    "Marks": [70, 80, 65, 90, 85, 75, 88, 92, 60, 95]
}

df = pd.DataFrame(data)

plt.scatter(df["Age"], df["Marks"])
plt.xlabel("Age")
plt.ylabel("Marks")
plt.title("Age vs Marks")
plt.show()
```
