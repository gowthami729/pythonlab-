#1. Number of Duplicates in a List

```python
a = [10, 20, 10, 30, 20, 10]

count = 0

for i in set(a):
    if a.count(i) > 1:
        count += 1

print("Duplicate elements =", count)
```

#Concatenate Tuples

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

t3 = t1 + t2

print("Concatenated tuple =", t3)
```

#Check Key Exists in Dictionary

```python
d = {"name": "Ravi", "age": 20, "marks": 85}

key = input("Enter key: ")

if key in d:
    print("Key exists")
else:
    print("Key does not exist")
```

#Sum of Dictionary Values

```python
d = {"a": 10, "b": 20, "c": 30}

total = sum(d.values())

print("Sum =", total)
```

#Dictionary Comprehension

```python
d = {x: x * x for x in range(1, 6)}

print(d)
```

#Frequency of Elements

```python
a = [1, 2, 2, 3, 1, 2, 4, 3]

freq = {}

for i in a:
    freq[i] = freq.get(i, 0) + 1

print(freq)
```
