#Add Two Matrices


```python
a = [[1, 2], [3, 4]]
b = [[5, 6], [7, 8]]

c = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        c[i][j] = a[i][j] + b[i][j]

print("Addition of matrices:")
for row in c:
    print(row)
```

#Multiply Two Matrices

```python
a = [[1, 2], [3, 4]]
b = [[5, 6], [7, 8]]

c = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            c[i][j] += a[i][k] * b[k][j]

print("Multiplication of matrices:")
for row in c:
    print(row)
```

#Transpose of a Matrix

```python
a = [[1, 2, 3], [4, 5, 6]]

t = [[0, 0], [0, 0], [0, 0]]

for i in range(2):
    for j in range(3):
        t[j][i] = a[i][j]

print("Transpose of matrix:")
for row in t:
    print(row)
```

#Class to Calculate Area and Perimeter of Circle, Triangle and Square

```python
import math

class Shape:

    def circle(self, r):
        print("Circle Area =", math.pi * r * r)
        print("Circle Perimeter =", 2 * math.pi * r)

    def square(self, s):
        print("Square Area =", s * s)
        print("Square Perimeter =", 4 * s)

    def triangle(self, a, b, c):
        p = a + b + c
        s = p / 2
        area = math.sqrt(s * (s - a) * (s - b) * (s - c))
        print("Triangle Area =", area)
        print("Triangle Perimeter =", p)

obj = Shape()

obj.circle(5)
obj.square(4)
obj.triangle(3, 4, 5)
```
