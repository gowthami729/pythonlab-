
#NumPy Array Creation and Properties
---



```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(a)
print("Dimensions =", a.ndim)
print("Shape =", a.shape)
print("Size =", a.size)
print("Data type =", a.dtype)
```

#Basic Slicing

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50, 60])

print(a[1:4])
print(a[:3])
print(a[2:])
print(a[::-1])
```

#Integer Indexing

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(a[0])
print(a[2])
print(a[-1])
```

#Boolean Indexing

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50, 60])

print(a[a > 30])
print(a[a % 20 == 0])
```

#Find Minimum, Maximum and Cumulative Sum

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Minimum =", np.min(a))
print("Maximum =", np.max(a))
print("Cumulative Sum =", np.cumsum(a))
```

#Convert Dictionary into Pandas DataFrame

```python
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Age": [18, 19, 20, 18, 21, 19, 20, 21, 18, 22],
    "Marks": [70, 80, 65, 90, 85, 75, 88, 92, 60, 95],
    "Height": [160, 165, 170, 155, 175, 168, 172, 180, 158, 176],
    "City": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"]
}

df = pd.DataFrame(data)

print(df)
```

#Select Any Two Columns and Observe Their Relationship

```python
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Age": [18, 19, 20, 18, 21, 19, 20, 21, 18, 22],
    "Marks": [70, 80, 65, 90, 85, 75, 88, 92, 60, 95],
    "Height": [160, 165, 170, 155, 175, 168, 172, 180, 158, 176],
    "City": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"]
}

df = pd.DataFrame(data)

selected = df[["Age", "Marks"]]

print(selected)
print(selected.corr())
```
