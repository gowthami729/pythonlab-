1. Arithmetic Operations


```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Modulus =", a % b)
print("Floor Division =", a // b)
print("Power =", a ** b)
```

2. Built-in Functions

```python
a = [10, 20, 30, 40, 50]

print("Length =", len(a))
print("Maximum =", max(a))
print("Minimum =", min(a))
print("Sum =", sum(a))
print("Sorted =", sorted(a))
```

3. List Methods

```python
a = [10, 20, 30]

a.append(40)
print(a)

a.insert(1, 15)
print(a)

a.remove(20)
print(a)

a.pop()
print(a)

a.sort()
print(a)

a.reverse()
print(a)
```

4. Length of a String

```python
s = input("Enter a string: ")
print("Length =", len(s))
```

5. Check Substring Present in a String

```python
s = input("Enter a string: ")
sub = input("Enter substring: ")

if sub in s:
    print("Substring is present")
else:
    print("Substring is not present")
```

6. Count Vowels

```python
s = input("Enter a string: ").lower()
count = 0

for ch in s:
    if ch in "aeiou":
        count += 1

print("Number of vowels =", count)
```

7. Anagram

```python
s1 = input("Enter first string: ").lower().replace(" ", "")
s2 = input("Enter second string: ").lower().replace(" ", "")

if sorted(s1) == sorted(s2):
    print("Anagram")
else:
    print("Not Anagram")
```
