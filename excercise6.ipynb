#File Handling – Read Mode

```python
f = open("sample.txt", "w")
f.write("Hello Python\nWelcome to File Handling")
f.close()
```

#File Handling – Write Mode

```python
f = open("sample.txt", "r")
data = f.read()
print(data)
f.close()
```

#File Handling – Append Mode

```python
f = open("sample.txt", "a")
f.write("\nNew line added")
f.close()
```

#Print Each Line of a File in Reverse Order

```python
f = open("sample.txt", "r")
lines = f.readlines()
f.close()

for line in reversed(lines):
    print(line.strip())
```

#Count Number of Characters in a File

```python
f = open("sample.txt", "r")
data = f.read()
f.close()

print("Characters =", len(data))
```

#Count Number of Lines in a File

```python
f = open("sample.txt", "r")
lines = f.readlines()
f.close()

print("Lines =", len(lines))
```
